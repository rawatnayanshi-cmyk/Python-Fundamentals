# DATA VISUALIZATION

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme (style="whitegrid",palette="pastel")
```

#  Total Revenue by Product Category (Bar Chart)

```python
df.groupby("product_category")["Revenue"].sum().plot(kind="bar", color="brown")
plt.title("Total Revenue by Product Category")
plt.ylabel("Revenue")
plt.show()
```

#  Average Revenue by Product Category (Bar Chart)

```python
df.groupby("product_category")["Revenue"].mean().plot(kind="bar", color="brown")
plt.title("Average Revenue by Product Category")
plt.ylabel("Average Revenue")
plt.show()
```

#  Count of Transactions by Product Category (Bar Chart)

```python
df["product_category"].value_counts().plot(kind="bar", color="teal")
plt.title("Transaction Count by Product Category")
plt.ylabel("Count")
plt.show()
```

# Revenue Trend Over Months (Line Chart)

```python
df.groupby("Month")["Revenue"].sum().plot(kind="line", marker="o", color="green")
plt.title("Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.show()
```

#  Revenue by Weekday (Bar Chart)

```python
df.groupby("Weekday")["Revenue"].sum().plot(kind="bar", color="purple")
plt.title("Revenue by Weekday")
plt.ylabel("Revenue")
plt.show()
```

# Revenue by Hour of Day (Line Chart)

```python
df.groupby("Hour")["Revenue"].sum().plot(kind="line", marker="o", color="red")
plt.title("Revenue by Hour of Day")
plt.ylabel("Revenue")
plt.show()
```

# . Revenue by Store Location (Pie Chart)

```python
df.groupby("store_location")["Revenue"].sum().plot(kind="pie", autopct="%1.1f%%")
plt.title("Revenue Share by Store Location")
plt.ylabel("")
plt.show()
```

#  Top 10 Product Details by Revenue (Horizontal Bar Chart)

```python
top_products = df.groupby("product_detail")["Revenue"].sum().nlargest(10)
top_products.plot(kind="barh", color="darkblue")
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.show()
```
