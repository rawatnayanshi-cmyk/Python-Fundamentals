# DATA VISUALIZATION

```python
sns.set_style("darkgrid")   
sns.set_palette("pastel")  
```

# Histogram: Distribution of release years

```python
plt.hist(df["release_year"], bins=30, color="skyblue", edgecolor="black")
plt.title("Distribution of Release Years")
plt.xlabel("Year")
plt.ylabel("Count")
```

# Countplot: Number of Movies vs TV Shows

```python
sns.countplot(x="type", data=df,hue= "type")
plt.title("Movies vs TV Shows")
plt.show()
```

# Barplot: Average duration by type

```python
sns.barplot(x="type", y="duration", data=df, hue="type")
plt.title("Average Duration by Type")
plt.show()
```

# Barplot: Top 10 countries by number of titles

```python
top_countries = df["country"].value_counts().nlargest(10)
sns.barplot(x=top_countries.index, y=top_countries.values)
plt.title("Top 10 Countries by Titles")
plt.xticks(rotation=45)
plt.show()
```

# Barplot

```python
top_genres = df["genres"].value_counts().nlargest(10)
sns.barplot(x=top_genres.index,y=top_genres.values)
plt.title("Top 10 Genres")
plt.xticks(rotation=45)
plt.show()
```

# Countplot: Ratings distribution

```python
sns.countplot(x="rating", data=df, hue="rating",order=df["rating"].value_counts().index)
plt.title("Ratings Distribution")
plt.xticks(rotation=45)
plt.show()
```

# Scatterplot: Duration vs Release Year

```python
sns.scatterplot(x="release_year", y="duration", hue="type", data=df)
plt.title("Duration vs Release Year")
plt.show()
```

# Heatmap: Correlation between numeric columns

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="plasma")
plt.title("Correlation Heatmap")
plt.show()
```
