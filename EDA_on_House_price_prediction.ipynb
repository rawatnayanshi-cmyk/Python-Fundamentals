**Data visualization**

```python
sns.histplot(df['price'], bins=50, kde=True, color='skyblue')
plt.title("Distribution of House Prices")
plt.show()
```

```python
sns.boxplot(x='bedrooms', y='price', data=df, palette='Set2')
plt.title("Price vs Bedrooms")
plt.show()
```

```python
sns.scatterplot(x='sqft_living', y='price', hue='waterfront', data=df, palette='coolwarm')
plt.title("Price vs Living Area (Waterfront Highlight)")
plt.show()
```

```python
df.groupby('bathrooms')['price'].mean().plot(kind='bar', color='teal')
plt.title("Average Price by Bathrooms")
plt.show()
```

```python
sns.boxplot(x='floors', y='price', data=df, palette='pastel')
plt.title("Price vs Floors")
plt.show()
```

```python
df['date'] = pd.to_datetime(df['date'])
df.groupby(df['date'].dt.to_period('M'))['price'].mean().plot(kind='line', color='purple')
plt.title("Average Price Trend Over Time")
plt.show()
```

```python
sns.heatmap(df[['price','sqft_living','sqft_lot','bedrooms','bathrooms']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
