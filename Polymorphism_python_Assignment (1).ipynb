```python
class flower:
    def color(self):
        print("Pink")
class plant:
    def color(self):
        print("Green")
# Common function
def colors(plant):
    plant.color()
#Demonstration 
flower = flower()
plant = plant()
colors(flower)
colors(plant)
```

```python
class vehicle:
    def drive(self):
        print("Vehicle is driving")
class car(vehicle):
    def drive(self):
        print("Car is driving")
class bike(vehicle):
    def drive(self):
        print("Bike is riding")
v = vehicle()
c = car()
b = bike()
v.drive()
c.drive()
b.drive()
```

```python
class box:
    def __init__(self,length,width,height):
        self.length = length 
        self.width = width 
        self.height = height
    def volume(self):
        return self.length * self.width * self.height
    def __add__ (self,other):
        return self.volume() + other.volume()
#Demostration 
box1 = box(2,3,4)
box2 = box(5,6,7)
print("Box1 volume:",box1.volume())
print("Box2 volume:",box2.volume())
print("Combined volume:",box1+box2)
```

```python
class calculator:
    def add(self,a,b):
        print("sum",a+b)
class calculator1:
    def add(self,a,b,c):
        print("sum",a+b+c)
#Demonstration
c = calculator()
c1 = calculator1()
c.add(10,20)
c1.add(10,20,30)
```
