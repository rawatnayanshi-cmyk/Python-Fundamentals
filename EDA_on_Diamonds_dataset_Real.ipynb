# Data visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme (style="whitegrid",palette="pastel")
```

# price distribution of carat

```python
sns.histplot(data= df,x="carat",bins=10)
```

# Number of diamonds by cut

```python
sns.countplot(data=df,x='cut')
```

# Relation Between price,carat and cut

```python
sns.scatterplot(data=df,x='carat',y='price',hue='cut') 
```

# Price Comparison Across cut

```python
sns.boxplot(data=df,x='cut',y='price')
```

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

# Correlation Heatmap

```python
sns.heatmap(data=corr_matrix,annot= True,fmt = '.2f')
```

# Number of Diamonds by color

```python
sns.countplot(x='color', data=df)
plt.show()
```

# Distribution of Diamonds by cut

```python
cut_counts = df['cut'].value_counts()

plt.pie(
    cut_counts,
    labels=cut_counts.index,
    autopct="%1.1f%%",  
    colors=sns.color_palette("pastel"),
    startangle=90
)
plt.title("Distribution of Diamond Cuts")
plt.show()
```
