```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    def show_brand(self):
        print("Vehicle Brand:", self.brand)
class Car(Vehicle):   # inherits from Vehicle
    def __init__(self, brand, model):
        super().__init__(brand)   # copy brand from Vehicle
        self.model = model
    def show_model(self):
        print("Car Model:", self.model)
# Object of Car
car1 = Car("Toyota", "Corolla")
car1.show_brand()
car1.show_model()
```

```python
class Sports:
    def play_sport(self):
        print("Playing Football")
class Music:
    def play_music(self):
        print("Playing Guitar")
class Student(Sports, Music):  
    def study(self):
        print("Studying Python")
# Object of Student
s1 = Student()
s1.play_sport()
s1.play_music()
s1.study()
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    def show_brand(self):
        print("Vehicle Brand:", self.brand)
class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model
    def show_model(self):
        print("Car Model:", self.model)
class ElectricCar(Car):
    def __init__(self, brand, model, battery):
        super().__init__(brand, model)
        self.battery = battery
    def show_battery(self):
        print("Battery Capacity:", self.battery, "kWh")
# Object of ElectricCar
ecar = ElectricCar("Tesla", "Model 3", 75)
ecar.show_brand()
ecar.show_model()
ecar.show_battery()
```

```python
class Shape:
    def draw(self):
        print("Drawing Shape")
class Circle(Shape):
    def show_name(self):
        print("Shape: Circle")
class Rectangle(Shape):
    def show_name(self):
        print("Shape: Rectangle")
c = Circle()
r = Rectangle()
c.draw()
c.show_name()
r.draw()
r.show_name()
```

```python
class Device:
    def show_device(self):
        print("This is a Device")
class Laptop(Device):
    def laptop_feature(self):
        print("Laptop Feature: Portable Computer")
class Tablet(Device):
    def tablet_feature(self):
        print("Tablet Feature: Touchscreen Device")
class HybridDevice(Laptop, Tablet):   # inherits from both Laptop & Tablet
    def hybrid_feature(self):
        print("Hybrid Feature: Laptop + Tablet")
hd = HybridDevice()
hd.show_device()
hd.laptop_feature()
hd.tablet_feature()
hd.hybrid_feature()
```

```python
class Employee:
    def show_employee(self):
        print("I am an Employee")
class Manager(Employee):
    def role(self):
        print("Role: Manager")
class Developer(Employee):
    def role(self):
        print("Role: Developer")
m = Manager()
d = Developer()
m.show_employee()
m.role()
d.show_employee()
d.role()
```
