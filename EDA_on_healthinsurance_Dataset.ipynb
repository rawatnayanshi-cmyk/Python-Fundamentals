**Data Visualization**

```python
sns.histplot(df['age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Gender Split")
plt.show()
```

```python
sns.scatterplot(x='age', y='bmi', hue='gender', data=df)
plt.title("BMI vs Age by Gender")
plt.show()

```

```python
sns.boxplot(x='children', y='bmi', data=df)
plt.title("BMI vs Children")
plt.show()
```

```python
sns.scatterplot(x='expenses', y='premium', hue='region', data=df)
plt.title("Expenses vs Premium by Region")
plt.show()
```

```python
sns.boxplot(x='region', y='expenses', data=df)
plt.title("Expenses by Region")
plt.show()
```

```python
sns.boxplot(x='discount_eligibility', y='premium', data=df)
plt.title("Premium by Discount Eligibility")
plt.show()
```

```python
df.groupby('region')['premium'].mean().plot(kind='bar',color='m')
plt.title("Average Premium per Region")
plt.show()
```
