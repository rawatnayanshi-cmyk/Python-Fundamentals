# DATA VISUALIZATION

```python
sns.set_style("darkgrid")   
sns.set_palette("pastel")  
```

# Histplot

```python
sns.histplot(df['total_bill'], kde=True)
plt.show()
```

```python
sns.histplot(data=df, x="total_bill", hue="day", multiple="stack")
```

```python
sns.histplot(data=df, x="total_bill", hue="sex")
```

# Countplot

```python
sns.countplot(x='day', data=df, hue= 'sex')
plt.title("Number of records per day")
plt.show()
```

# Scatterplot

```python
sns.scatterplot(x='total_bill', y='tip', data=df, hue = 'sex',style="smoker")
plt.title("Total bill vs Tip")
plt.show()
```

 # Boxplot

```python
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

# Barplot

```python
sns.barplot(x='sex', y='tip', data=df,hue ="sex",palette="pastel")
plt.show()
```

# Heatmap

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="plasma")
plt.title("Correlation Heatmap")
plt.show()
```

# Piechart

```python
smoker_counts = df['smoker'].value_counts()
plt.pie(
    smoker_counts,
    labels=smoker_counts.index,
    autopct="%1.1f%%",   
    colors=["lightcoral", "lightblue"],
    startangle=90
)
plt.title("Smokers vs Non-Smokers")
plt.show()
```

# Pairplot

```python
sns.pairplot(df,hue='day')
plt.show()
```
