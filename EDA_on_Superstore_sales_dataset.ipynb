# DATA VISUALIZATION

```python
sns.set_style("darkgrid")   
sns.set_palette("pastel")  
```

# Barplot

```python
sns.barplot(x="Category", y="Sales", data=df, estimator=sum,hue="Category")
plt.title("Total Sales by Category")
plt.show()
```

#  Count plot: Number of orders by Ship Mode

```python
sns.countplot(x="Ship Mode", data=df,hue="Ship Mode")
plt.title("Order Count by Ship Mode")
plt.show()
```

# Box plot: Sales distribution by Region

```python
sns.boxplot(x="Region", y="Sales", data=df)
plt.title("Sales Distribution by Region")
plt.show()
```

#  Heatmap: Correlation between numeric columns

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="plasma")
plt.title("Correlation Heatmap")
plt.show()
```

# Pie chart: Sales share by Region

```python
sales_by_region = df.groupby("Region")["Sales"].sum()
plt.pie(sales_by_region, labels=sales_by_region.index, autopct="%1.1f%%")
plt.title("Sales Share by Region")
plt.show()
```


#  Histogram: Distribution of Sales

```python
plt.hist(df["Sales"], bins=30, color="skyblue", edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

# Stacked bar chart

```python
sales_cat_region = df.groupby(["Category", "Region"])["Sales"].sum().unstack()
sales_cat_region.plot(kind="bar", stacked=True)
plt.title("Sales by Category and Region")
plt.ylabel("Sales")
plt.show()
```
