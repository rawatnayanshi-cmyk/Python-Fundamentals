# DATA VISUALIZATION

```python
sns.set_style("darkgrid")   
sns.set_palette("pastel")  
```

# Histplot

```python
sns.histplot(df['alcohol'], kde=True, color="skyblue")
plt.title("Alcohol Distribution")
plt.show()
```

# Countplot

```python
sns.countplot(x="quality", data=df, palette="pastel",hue = "quality")
plt.title("Wine Quality Counts")
plt.show()
```

# Scatterplot

```python
sns.scatterplot(x="alcohol", y="quality", data=df,hue="quality")
plt.title("Alcohol vs Quality")
plt.show()
```

# Boxplot

```python
sns.boxplot(x="quality", y="volatile acidity", data=df, palette="muted",hue="quality")
plt.title("Volatile Acidity vs Quality")
plt.show()
```

# Heatmap

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="plasma")
plt.title("Correlation Heatmap")
plt.show()
```
