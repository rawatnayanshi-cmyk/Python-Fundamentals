# DATA VISUALIZATION

```python
sns.set_style("darkgrid")   
sns.set_palette("pastel")  
```

# Distribution of Scores

```python
sns.histplot(df['math_score'], kde=True, color="skyblue")
plt.title("Math Score Distribution")
plt.show()
```

```python
sns.countplot(x="final_grade", data=df, palette="pastel",hue="final_grade")
plt.title("Final Grade Distribution")
plt.show()
```

# Gender vs scores

```python
sns.boxplot(x="gender", y="overall_score", data=df, palette="pastel",hue="gender")
plt.title("Overall Score by Gender")
plt.show()
```

# School Type vs Study Hours

```python
sns.barplot(x="school_type", y="study_hours", data=df, palette="pastel",hue="school_type")
plt.title("Average Study Hours by School Type")
plt.show()
```

# Study Hours vs Overall Score

```python
sns.scatterplot(x="study_hours", y="overall_score", data=df, hue="gender")
plt.title("Study Hours vs Overall Score")
plt.show()
```

# Attendance vs Final Grade

```python
sns.boxplot(x="final_grade", y="attendance_percentage", data=df, palette="coolwarm",hue="final_grade")
plt.title("Attendance vs Final Grade")
plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="plasma")
plt.title("Correlation Heatmap")
plt.show()
```
