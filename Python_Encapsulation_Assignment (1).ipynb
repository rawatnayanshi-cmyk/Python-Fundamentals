```python
class Employee:
    def __init__(self):
        self.company = "Microsoft"   
emp = Employee()
print(emp.company)
```

```python
class Car:
    def __init__(self):
        self._model = "BMW"  
c = Car()
print(c._model)   
```

```python
class Book:
    def __init__(self):
        self.__price = 700
    def display(self):
        print(self.__price)
book = Book()
book.display()
```

```python
class Laptop:
    def __init__(self):
        self.__brand = "Lenovo"
    def get_brand(self):   # getter
        return self.__brand
l = Laptop()
print(l.get_brand())
```

```python
class Mobile:
    def __init__(self):
        self.__battery = 100
    def set_battery(self, value):   # setter
        self.__battery = value
    def get_battery(self):   # getter
        return self.__battery
m = Mobile()
m.set_battery(90)
print("Battery:", m.get_battery())
```

```python
class Product:
    def __init__(self):
        self.__quantity = 10
    def set_quantity(self, value):
        if value >= 0:
            self.__quantity = value
        else:
            print("Invalid Quantity")
    def get_quantity(self):
        return self.__quantity
p = Product()
p.set_quantity(10)
print("Quantity:", p.get_quantity())
p.set_quantity(-5)  
```

```python
class Bank:
    def __init__(self):
        self.__balance = 2000
    def deposit(self, amount):
        self.__balance += amount
    def get_balance(self):   # getter
        return self.__balance
b = Bank()
b.deposit(1000)
print("Balance:", b.get_balance())
```

```python
class Exam:
    def __init__(self):
        self.__marks = 0
    def set_marks(self, value):
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Marks should be between 0 and 100")
    def get_marks(self):
        return self.__marks
e = Exam()
e.set_marks(80)
print("Marks:", e.get_marks())
e.set_marks(120)  
```

```python
class Person:
    def __init__(self, name, age, salary):
        self.name = name          
        self._age = age           
        self.__salary = salary  
    def set_salary(self, value): 
        self.__salary = value
    def get_salary(self):          
        return self.__salary
p = Person("Nitya", 25, 50000)
print("Name:", p.name)      
print("Age:", p._age)        
print("Salary:", p.get_salary())  
```

```python
class LibraryBook:
    def __init__(self):
        self.__title = ""
        self.__author = ""
    def set_title(self, title):
        self.__title = title
    def set_author(self, author):
        self.__author = author
    def get_title(self):
        return self.__title
    def get_author(self):
        return self.__author
book = LibraryBook()
book.set_title("Python Basics")
book.set_author("John Smith")
print("Title:", book.get_title())
print("Author:", book.get_author())
```
