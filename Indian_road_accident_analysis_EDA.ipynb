# Data Visualization

```python
sns.set_palette("pastel")
sns.set_style("whitegrid")   
```

#  Accidents by Day of Week

```python
sns.countplot(x="day_of_week", data=df, order=df["day_of_week"].value_counts().index)
plt.title("Accidents by Day of Week")
plt.show()
```

#  Weekend vs Weekday Accidents

```python
sns.countplot(x="is_weekend", data=df)
plt.title("Weekend vs Weekday Accidents")
plt.show()
```

# Hourly Accident Distribution


```python
sns.histplot(df["hour"], bins=24, kde=False)
plt.title("Accidents by Hour of Day")
plt.show()
```

# Accidents by Road Type

```python
sns.countplot(y="road_type", data=df, order=df["road_type"].value_counts().index)
plt.title("Accidents by Road Type")
plt.show()
```

#  Weather vs Accident Severity

```python
sns.countplot(x="weather", hue="accident_severity", data=df)
plt.title("Weather vs Accident Severity")
plt.xticks(rotation=45)
plt.show()
```

#  Visibility vs Risk Score

```python
sns.boxplot(x="visibility", y="risk_score", data=df)
plt.title("Risk Score by Visibility")
plt.show()
```

# Traffic Density vs Casualties

```python
sns.barplot(x="traffic_density", y="casualties", data=df, estimator="sum")
plt.title("Total Casualties by Traffic Density")
plt.show()
```

 # Vehicles Involved vs Severity

```python
sns.countplot(x="vehicles_involved", hue="accident_severity", data=df)
plt.title("Vehicles Involved vs Accident Severity")
plt.show()
```

#  Causes of Accidents

```python
top_causes = df["cause"].value_counts().nlargest(10)
sns.barplot(x=top_causes.index, y=top_causes.values)
plt.title("Top 10 Causes of Accidents")
plt.xticks(rotation=45)
plt.show()
```

#  Correlation Heatmap (numeric features)

```python
plt.figure(figsize=(10,6))
sns.heatmap(df[["hour","lanes","temperature","vehicles_involved","casualties","risk_score"]].corr(),
            annot=True, cmap="plasma")
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```
