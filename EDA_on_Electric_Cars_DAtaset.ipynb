# Data Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme (style="whitegrid",palette="pastel")
```

```python
# 1. Distribution of Car Prices
plt.figure(figsize=(8,5))
sns.histplot(df["PriceEuro"], bins=30, kde=True)
plt.title("Distribution of Electric Car Prices")
plt.show()
```

```python

# 2. Top 10 Brands by Number of Models
plt.figure(figsize=(10,5))
df["Brand"].value_counts().head(10).plot(kind="bar", color="skyblue")
plt.title("Top 10 Brands by Number of Models")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Range vs Price Scatter Plot
plt.figure(figsize=(8,6))
sns.scatterplot(x="Range_Km", y="PriceEuro", hue="Brand", data=df)
plt.title("Range vs Price of Electric Cars")
plt.show()
```

```python
# 4. Acceleration vs Top Speed
plt.figure(figsize=(8,6))
sns.scatterplot(x="AccelSec", y="TopSpeed_KmH", hue="Brand", data=df)
plt.title("Acceleration vs Top Speed")
plt.show()
```

```python
# 5. Average Range by Body Style
plt.figure(figsize=(8,6))
sns.barplot(x="BodyStyle", y="Range_Km", data=df, estimator="mean", ci=None)
plt.title("Average Range by Body Style")
plt.show()
```

```python
# 6. Efficiency Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Efficiency_WhKm"], bins=20, kde=True, color="orange")
plt.title("Distribution of Efficiency (Wh/km)")
plt.show()
```

```python
# 7. Seats vs Price Boxplot
plt.figure(figsize=(8,6))
sns.boxplot(x="Seats", y="PriceEuro", data=df)
plt.title("Price Distribution by Number of Seats")
plt.show()
```

```python
# 9. Segment vs Price
plt.figure(figsize=(8,6))
sns.boxplot(x="Segment", y="PriceEuro", data=df)
plt.title("Price Distribution by Segment")
plt.show()
```

```python
# 10. Range Comparison of Tesla Models
plt.figure(figsize=(10,6))
tesla = df[df["Brand"].str.contains("Tesla")]
sns.barplot(x="Model", y="Range_Km", data=tesla, ci=None)
plt.title("Range Comparison of Tesla Models")
plt.xticks(rotation=45)
plt.show()
```
