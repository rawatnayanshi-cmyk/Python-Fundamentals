# Data Visualization

```python
sns.set_theme(style="whitegrid",palette="pastel")
```

# Price Distribution of products

```python
sns.histplot(data=df,x="Price_USD",kde = True ,bins=10)
```

# Price comparison Across Categories

```python
sns.boxplot(x="Category", y="Price_USD", data=df)
plt.title("Price Distribution by Category")
plt.xticks(rotation=45)
plt.show()
```

# Relationship Between price and rating

```python
sns.scatterplot(x="Price_USD", y="Rating", data=df)
plt.title("Price vs Rating")
plt.show()
```

# Pairwise comparison of price,rating and number of reviews

```python
sns.pairplot(df[["Price_USD", "Rating", "Number_of_Reviews"]])
plt.suptitle("Pairplot of Numeric Columns", y=1.02)
plt.show()
```

# Average Price per category

```python
sns.barplot(x="Category", y="Price_USD", data=df, estimator="mean")
plt.title("Average Price by Category")
plt.xticks(rotation=45)
plt.show()
```

# Gender Distribution

```python
df["Gender_Target"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```
